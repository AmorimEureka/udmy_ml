```python
n = 18
```

```python
import math as mt

raiz = mt.sqrt(n)

print(f'A raiz quadrada de {n} é {raiz:.2f}')
```

```python
nw_raiz = n ** 0.5
print(f'A nova raiz quadrada pode ser obtida pela expessão acima e retorna o mesmo resultado: {nw_raiz:.3f}')
```
